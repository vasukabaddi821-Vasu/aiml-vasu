```python
from pypdf import PdfReader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Read PDF
reader = PdfReader(pdf_path)

text = ""
for page in reader.pages:
    text += page.extract_text() or ""

# Split text into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_text(text)

print("PDF text extracted:", len(text), "characters")
print("Chunks created:", len(chunks))
print("First chunk:\n", chunks[0])
```

```python
def retrieve(question, top_k=3):

    # Convert question into embedding
    question_embedding = embedding_model.encode(
        [question],
        normalize_embeddings=True
    ).astype("float32")

    # Search FAISS for similar chunks
    scores, indices = index.search(question_embedding, top_k)

    # Get matching text chunks
    results = []

    for i in indices[0]:
        if i >= 0:
            results.append(chunks[i])

    return "\n\n".join(results)


question = input("Ask a question about your PDF: ")

context = retrieve(question)

print("\nRelevant information found:\n")
print(context)
```

```python
from huggingface_hub import InferenceClient
from getpass import getpass

HF_TOKEN = getpass("Paste your Hugging Face token: ")

llm = InferenceClient(
    api_key=HF_TOKEN
)

print("Llama connection client initialized!")
```

```python
def rag_answer(question):

    # Retrieve relevant chunks from PDF
    context = retrieve(question, top_k=3)

    prompt = f"""
You are an AI assistant answering questions about an uploaded PDF.

Use only the context provided below.
If the answer is not available in the context,
say that you could not find it in the document.

CONTEXT:
{context}

QUESTION:
{question}

ANSWER:
"""

    response = llm.chat_completion(
        model="meta-llama/Llama-3.1-8B-Instruct",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_tokens=300,
        temperature=0.2
    )

    return response.choices[0].message.content
```

```python
import gradio as gr
from pypdf import PdfReader
from langchain_text_splitters import RecursiveCharacterTextSplitter
import numpy as np
import faiss

# Process the uploaded PDF
def process_pdf(file):
    if file is None:
        return None, None, "Please upload a PDF first."

    try:
        reader = PdfReader(file)

        pdf_text = ""
        for page in reader.pages:
            pdf_text += page.extract_text() or ""

        if not pdf_text.strip():
            return None, None, "No readable text found in this PDF."

        splitter = RecursiveCharacterTextSplitter(
            chunk_size=500,
            chunk_overlap=100
        )

        document_chunks = splitter.split_text(pdf_text)

        vectors = embedding_model.encode(
            document_chunks,
            show_progress_bar=False,
            normalize_embeddings=True
        )

        vectors = np.asarray(vectors, dtype="float32")

        vector_index = faiss.IndexFlatIP(
            vectors.shape[1]
        )

        vector_index.add(vectors)

        status = (
            f"✅ PDF processed successfully!\n"
            f"Pages: {len(reader.pages)}\n"
            f"Chunks: {len(document_chunks)}"
        )

        return document_chunks, vector_index, status

    except Exception as e:
        return None, None, f"Error processing PDF: {e}"


# Answer questions using the uploaded PDF
def ask_pdf(question, document_chunks, vector_index):
    if not question.strip():
        return "Please enter a question."

    if document_chunks is None or vector_index is None:
        return "Please upload and process a PDF first."

    try:
        question_vector = embedding_model.encode(
            [question],
            normalize_embeddings=True
        ).astype("float32")

        scores, indices = vector_index.search(
            question_vector, 3
        )

        retrieved_chunks = []

        for idx in indices[0]:
            if idx >= 0:
                retrieved_chunks.append(
                    document_chunks[idx]
                )

        context = "\n\n".join(retrieved_chunks)

        prompt = f"""
You are a document question-answering assistant.

Answer the user's question using only the provided document context.
Explain clearly in simple words.
If the information is not present, say that it is not found in the document.

DOCUMENT CONTEXT:
{context}

QUESTION:
{question}

ANSWER:
"""

        response = client.models.generate_content(
            model=MODEL,
            contents=prompt
        )

        answer = response.text or "No answer was generated."

        source_text = "\n\n--- Retrieved Sources ---\n"

        for i, chunk in enumerate(retrieved_chunks, 1):
            source_text += f"\nSource {i}:\n{chunk}\n"

        return answer + source_text

    except Exception as e:
        return f"Error answering question: {e}"
```

```python
def ask_pdf_chat(question, history, document_chunks, vector_index):
    if not question.strip():
        return history, ""

    if document_chunks is None or vector_index is None:
        history = history + [
            {"role": "assistant", "content": "Please upload and process a PDF first."}
        ]
        return history, ""

    try:
        question_vector = embedding_model.encode(
            [question],
            normalize_embeddings=True
        ).astype("float32")

        scores, indices = vector_index.search(question_vector, 3)

        retrieved_chunks = [
            document_chunks[idx]
            for idx in indices[0]
            if idx >= 0
        ]

        context = "\n\n".join(retrieved_chunks)

        history_text = ""

        for msg in history[-6:]:
            history_text += f"{msg['role']}: {msg['content']}\n"

        prompt = f"""
You are a conversational PDF assistant.

Answer using the document context.
Use conversation history to understand follow-up questions.
Explain in simple language.
If the answer is not in the document, clearly say so.

DOCUMENT CONTEXT:
{context}

CONVERSATION HISTORY:
{history_text}

CURRENT QUESTION:
{question}

ANSWER:
"""

        response = client.models.generate_content(
            model=MODEL,
            contents=prompt
        )

        answer = response.text or "No answer generated."

        history = history + [
            {"role": "user", "content": question},
            {"role": "assistant", "content": answer}
        ]

        return history, ""

    except Exception as e:
        history = history + [
            {"role": "assistant", "content": f"Error: {e}"}
        ]
        return history, ""
```

```python
with gr.Blocks(title="Document AI Chatbot") as demo:

    gr.Markdown("# 📚 Document AI Chatbot")
    gr.Markdown("Upload a PDF and ask questions about it.")

    document_state = gr.State()
    index_state = gr.State()

    pdf_upload = gr.File(
        label="Upload PDF",
        file_types=[".pdf"]
    )

    process_button = gr.Button("Upload & Process")

    status_box = gr.Textbox(label="Document Status")

    chatbot = gr.Chatbot(label="Conversation")

    question_box = gr.Textbox(
        label="Ask a question",
        placeholder="Ask about your PDF..."
    )

    ask_button = gr.Button("Send")

    process_button.click(
        process_pdf,
        inputs=pdf_upload,
        outputs=[document_state, index_state, status_box]
    )

    ask_button.click(
        ask_pdf_chat,
        inputs=[question_box, chatbot, document_state, index_state],
        outputs=[chatbot, question_box]
    )

    question_box.submit(
        ask_pdf_chat,
        inputs=[question_box, chatbot, document_state, index_state],
        outputs=[chatbot, question_box]
    )

demo.launch(share=True)
```
